```python
import pdfplumber
import pandas as pd
import re
from pathlib import Path

print("PDF extraction tools ready")
```

```python
sales_pdf = Path("../data/raw/sales_history.pdf")
inventory_pdf = Path("../data/raw/inventory_count.pdf")

print(sales_pdf.exists())
print(inventory_pdf.exists())
```

```python
with pdfplumber.open(sales_pdf) as pdf:
    first_page_text = pdf.pages[0].extract_text()

print(first_page_text[:2000])
```

```python
all_sales_text = ""

with pdfplumber.open(sales_pdf) as pdf:
    print("Number of pages:", len(pdf.pages))
    
    for page in pdf.pages:
        text = page.extract_text()
        if text:
            all_sales_text += text + "\n"

print(all_sales_text[:1000])
```

```python
sales_lines = all_sales_text.split("\n")

print("Total lines:", len(sales_lines))
print("First 20 lines:")
for line in sales_lines[:20]:
    print(line)
```

```python
transaction_lines = []

for line in sales_lines:
    line = line.strip()
    

    if re.match(r"^\d{6}\s+", line):
        transaction_lines.append(line)

print("Transaction lines found:", len(transaction_lines))
print("First 10 transaction lines:")

for line in transaction_lines[:10]:
    print(line)
```

```python
sales_rows = []

for line in transaction_lines:
    parts = line.split()
    
    tx_number = parts[0]
    inventory_number = parts[1]
    
    
    cases = parts[-9]
    kegs = parts[-8]
    other = parts[-7]
    price = parts[-6]
    cost = parts[-5]
    total_price = parts[-4]
    total_cost = parts[-3]
    gross_profit = parts[-2]
    category = parts[-1]
    

    description = " ".join(parts[2:-9])
    
    sales_rows.append({
        "tx_number": tx_number,
        "inventory_number": inventory_number,
        "description": description,
        "cases": cases,
        "kegs": kegs,
        "other": other,
        "price": price,
        "cost": cost,
        "total_price": total_price,
        "total_cost": total_cost,
        "gross_profit": gross_profit,
        "category": category
    })

sales_df = pd.DataFrame(sales_rows)

sales_df.head()
```

```python
numeric_columns = [
    "cases", "kegs", "other", "price", "cost",
    "total_price", "total_cost", "gross_profit"
]

for col in numeric_columns:
    sales_df[col] = pd.to_numeric(sales_df[col], errors="coerce")

sales_df["profit_margin"] = sales_df["gross_profit"] / sales_df["total_price"]

sales_df.head()
```

```python
print("Rows:", len(sales_df))
print("Total Revenue:", sales_df["total_price"].sum())
print("Total Gross Profit:", sales_df["gross_profit"].sum())
print("Total Cases:", sales_df["cases"].sum())

sales_df.describe()
```

```python
sales_rows = []

for line in transaction_lines:
    parts = line.split()

    tx_number = parts[0]
    inventory_number = parts[1]

    pd_period = parts[-1]
    category = parts[-2]
    gross_profit = parts[-3]
    total_cost = parts[-4]
    total_price = parts[-5]
    cost = parts[-6]
    price = parts[-7]

  
    before_price = parts[2:-7]


    quantity = before_price[-1]
    description = " ".join(before_price[:-1])

    sales_rows.append({
        "tx_number": tx_number,
        "inventory_number": inventory_number,
        "description": description,
        "quantity_sold": quantity,
        "price": price,
        "cost": cost,
        "total_price": total_price,
        "total_cost": total_cost,
        "gross_profit": gross_profit,
        "category": category,
        "period": pd_period
    })

sales_df = pd.DataFrame(sales_rows)

sales_df.head(10)
```

```python
numeric_cols = [
    "quantity_sold",
    "price",
    "cost",
    "total_price",
    "total_cost",
    "gross_profit"
]

for col in numeric_cols:
    sales_df[col] = pd.to_numeric(sales_df[col], errors="coerce")

sales_df["profit_margin"] = sales_df["gross_profit"] / sales_df["total_price"]

sales_df.head(10)
```

```python
print("Rows:", len(sales_df))
print("Total Revenue:", sales_df["total_price"].sum())
print("Total Gross Profit:", sales_df["gross_profit"].sum())
print("Total Quantity Sold:", sales_df["quantity_sold"].sum())

sales_df.describe()
```

```python
sales_df.to_csv("../exports/clean_sales.csv", index=False)

print("Sales data exported successfully!")
```
